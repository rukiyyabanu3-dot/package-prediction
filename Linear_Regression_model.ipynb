```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split as tts
from sklearn.linear_model import LinearRegression as lrr
data=pd.read_csv("C:/Users/CS-06/OneDrive/place.csv")
print(data)
```

```python
print(data.isnull().sum())
```

```python
x=data.iloc[:,[-2]].values
y=data.iloc[:,[-1]].values
print(x)
print(y)
```

```python
x_train,x_test,y_train,y_test=tts(x,y,test_size=0.2)
print(x_train,x_test,y_train,y_test)
```

```python
ls=lrr()
ls.fit(x_train,y_train)
```

```python
y_predict=ls.predict(x_test)
print(y_predict)
```

```python
train_accuracy=ls.score(x_train,y_train)
test_accuracy=ls.score(x_test,y_test)
print(train_accuracy)
print(test_accuracy)
```

```python
from sklearn.metrics import mean_squared_error
msc=mean_squared_error(y_test,y_predict)
rmsc=np.sqrt(msc)
print("MSE:", msc)
print("RMSE:", rmsc)
```
